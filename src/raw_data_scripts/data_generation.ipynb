```python
from faker import Faker
import random
from datetime import datetime
import csv

faker = Faker()

print("Name:", faker.name())
print("Address:", faker.address())
print("Email:", faker.email())
print("Phone Number:", faker.phone_number())
print("Job Title:", faker.job())
print("Date of Birth:", faker.date_of_birth())
print("Credit Card Number:", faker.credit_card_number())
```

```python

def generate_customers():
    num_of_countrie = 10000
    countries = ["BE", "FR", "DE", "NL", "ES"]

    with open("output/customers.csv", "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow([
            "customer_id",
            "first_name",
            "last_name",
            "email",
            "country",
            "signup_date"
        ])

        for customer_id in range(1, num_of_countrie + 1):
            writer.writerow([
                customer_id,
                faker.first_name(),
                faker.last_name(),
                faker.unique.email(),
                random.choice(countries),
                faker.date_between(start_date='-3y', end_date='today')
            ])
```

```python
def generate_products():
    NUM_PRODUCTS = 1000
    CATEGORIES = ["electronics", "fashion", "home", "sports", "books"]
    CURRENCIES = ["EUR"]

    with open("output/products.csv", "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow([
            "product_id",
            "product_name",
            "category",
            "price",
            #'description',
            "currency",
            "created_at"
        ])

        for product_id in range(1, NUM_PRODUCTS + 1):
            writer.writerow([
                product_id,
                faker.word().capitalize(),
                random.choice(CATEGORIES),
                round(random.uniform(10, 500), 2),
                #faker.sentence(nb_words=10),
                "EUR",
                faker.date_between(start_date="-2y", end_date="today")
            ])
```

```python
def generate_orders():
    NUM_ORDERS = 100000
    ORDER_STATUSES = ["completed", "cancelled", "refunded"]

    with open("output/orders.csv", "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow([
            "order_id",
            "customer_id",
            "order_date",
            "order_status",
            "currency"
        ])

        for order_id in range(1, NUM_ORDERS + 1):
            writer.writerow([
                order_id,
                random.randint(1, 10000),
                faker.date_between(start_date="-2y", end_date="today"),
                random.choices(
                    ORDER_STATUSES,
                    weights=[0.85, 0.1, 0.05]
                )[0],
                "EUR"
            ])
```

```python
def generate_order_items():
    NUM_ORDERS = 100000
    NUM_PRODUCTS = 1000

    order_item_id = 1

    with open("output/order_items.csv", "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow([
            "order_item_id",
            "order_id",
            "product_id",
            "quantity",
            "unit_price",
            "line_total"
        ])

        for order_id in range(1, NUM_ORDERS + 1):
            num_items = random.randint(1, 5)

            for _ in range(num_items):
                quantity = random.randint(1, 3)
                unit_price = round(random.uniform(10, 500), 2)

                writer.writerow([
                    order_item_id,
                    order_id,
                    random.randint(1, NUM_PRODUCTS),
                    quantity,
                    unit_price,
                    round(quantity * unit_price, 2)
                ])

                order_item_id += 1
```

```python
def generate_payments():
    PAYMENT_METHODS = ["credit_card", "paypal", "bank_transfer"]
    PAYMENT_STATUSES = ["success", "failed"]

    with open("output/payments.csv", "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow([
            "payment_id",
            "order_id",
            "payment_method",
            "payment_status",
            "payment_date",
            "amount"
        ])

        payment_id = 1

        for order_id in range(1, 100000 + 1):
            status = random.choices(
                PAYMENT_STATUSES,
                weights=[0.9, 0.1]
            )[0]

            writer.writerow([
                payment_id,
                order_id,
                random.choice(PAYMENT_METHODS),
                status,
                faker.date_between(start_date="-2y", end_date="today"),
                round(random.uniform(10, 1000), 2)
            ])

            payment_id += 1
```

```python
generate_customers()
generate_products()
generate_orders()
generate_order_items()
generate_payments()
```
